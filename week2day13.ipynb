```python
# 📌 Task 2 — Use a Built-in Module
# Create a script:
# Generate a random OTP using random module
# Print OTP like:
    # Your secure OTP is: 48216

# Push to GitHub ✅

import random

def generate_otp(length=5):
    start = 10**(length - 1)
    end = (10**length) - 1
    return random.randint(start, end)

otp = generate_otp()
print(f"🔐 Your secure OTP is: {otp}")
```
