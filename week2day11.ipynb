```python
# 🐍 Python Day 11 — Object-Oriented Programming (OOP)
# Class & Object

class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

stud1 = Student('Komal', 98)

print("Name: ", stud1.name)
print("Score: ", stud1.score)
```

```python
# Methods Inside Class

class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score
    
    def greet(self):
        print(f'Hello, {self.name}!')

stud1 = Student('Komal', 97)
stud1.greet()
```

```python
# Encapsulation - methods are used to access, update data

class Account:
    def __init__(self, holder, balance):
        self.holder = holder
        self.__balance = balance  # Private
    
    def deposit(self, amt):
        self.__balance += amt
    
    def get_balance(self):
        return self.__balance

acc = Account("Komal", 5000)
acc.deposit(1000)
print(acc.get_balance())
```

```python
# Inheritance

class Person:
    def __init__(self, name):
        self.name = name
    
    def intro(self):
        return f'Hi, I\'m {self.name}!'
    
class Student(Person):  # No need for constructor
    def study(self):
        return f'{self.name} is studying.'

stud1 = Student('Komal')
print(stud1.intro())
print(stud1.study())
```

```python
# Polymorphism

class Animal:
    def sound(self):
        return 'Animals make sound.'

class Dog(Animal):
    def sound(self):
        return 'Dogs bark.'

a = Animal()
d = Dog()

print(a.sound())
print(d.sound())
```

```python
# Data Science Style Example

class Model:
    def __init__(self, name):
        self.name = name
        self.is_trained = False
    
    def train(self, data):
        self.is_trained = True
        print(f'{self.name} trained on {len(data)} records.')
    
    def predict(self, x):
        if not self.is_trained:
            print("Train the model first!")
        else:
            return f'Predicted result for {x}.'

model = Model("Student Score Predictor")
model.train([1, 2, 3, 4, 5])
print(model.predict(90))
```

```python
# Day 11 Mini Project
# 📌 Product Billing System
# Requirements:
#  ✔ Class Product(name, price, quantity)
#  ✔ Method to print product details
#  ✔ Method to calculate total bill
#  ✔ Create 3 products → compute final bill
#  ✔ Print formatted invoice ✅
# Starter Code:
# class Product:
#     pass  # ✅ You will write full code here

# 🎯 Try to implement:
# __init__
# methods like get_total()
# store multiple products in a list
# loop to display bill properly

# Example Invoice Output:
# ------ Invoice ------
# Product | Price | Qty | Total
# Milk       40      2     80
# Eggs       60      1     60
# Bread      35      2     70
# -----------------------
# Grand Total = 210

# 📌 Then push to GitHub with commit:
# "Day 11 OOP Billing System ✅"
# Show me your code when done — I’ll review and upgrade it to portfolio quality ⭐

import json

class Product:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity
    
    def get_total(self):
        return self.price * self.quantity

products = []
flag = 'y'
no_of_products = 0

# Collecting products
while (flag == 'y' or flag == 'Y'):
    no_of_products += 1
    print(f'Enter details for product {no_of_products}.')
    name = input("Product name: ")
    price = float(input("Product price: "))
    quantity = int(input("Product quantity: "))
        
    item = Product(name, price, quantity)   # Create object
    products.append(item)   # Add object to the list
    
    flag = input("Do you want to add more products? (Y/N)")
    
    if flag not in 'yYnN':
        raise ValueError("Error! Choose (Y/N).")

grand_total = 0
print("\n----------- Invoice -----------")
print(f"Product | Price | Qty | Total")

for product in products:
    total = product.get_total()
    print(f'{product.name:10} | {product.price:5} | {product.quantity:3} | {total}')
    grand_total += total
        
print("-------------------------------")
print(f"Grand Total = {grand_total}")

invoice = [{"name": p.name, "price": p.price, "quantity": p.quantity, "total":p.get_total()} for p in products]

with open("invoice.json", 'w') as file:
    json.dump(invoice, file, indent = 4)

with open("invoice.json", 'r') as file:
    bill = json.load(file)
    bill.append({"grand total":grand_total})

print("\n✅ Invoice saved as invoice.json!")
```
