```python
# 🐍 Python Day 12 — Advanced OOP + Real Practice
# 1. Class vs Static Methods

class Student:
    def __init__(self, name, marks):
        self.name = name,
        self.marks = marks

    # Class Methods
    def display(self):
        return f'Marks: {self.marks}'
    
    # Static Methods
    def addition(a, b):
        return (a + b)

stud1 = Student('Komal', 97)
print(stud1.display())
print(Student.addition(5, 13))
```

```python
# 2. Magic / Dunder Methods

class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages
    
    def __str__(self):
        return f"{self.title} ({self.pages} pages)"
    
    def __len__(self):
        return self.pages

b = Book("Python Mastery", 350)
print(b)
print(len(b))
```

```python
# 3. Composition (HAS-A relationship)

class Engine:
    def __init__(self, power):
        self.power = power
    
class Car:
    def __init__(self, brand, engine):
        self.brand = brand
        self.engine = engine

e = Engine("200HP")
c = Car('Mercedez', e)
print(f'Car: {c.brand}\nEngine power: {c.engine.power}')
```

```python
# 4. Real Data Science Example - Dataset Loader Class

import csv

class DataLoader:
    def __init__(self, filepath):
        self.filepath = filepath
        self.data = []

    def load_csv(self):
        with open(self.filepath, "r") as file:
            reader = csv.DictReader(file)
            self.data = list(reader)
        return self.data
    
    def __len__(self):
        return len(self.data)

loader = DataLoader("users.csv")
loader.load_csv()
print(loader.data)
print("Records:", len(loader))
```

```python
# Day 12 Mini Projects (2 Projects!)
# Project 1: Student Management System (Portfolio-worthy)
# Requirements:
#  ✅ Store multiple students
#  ✅ Each student has marks in 3 subjects
#  ✅ Calculate total + average
#  ✅ Grade A/B/C based on average
#  ✅ Pretty formatted report

# Output Example:
# ----- Student Report -----
# Name: Komal
# Math: 88, Sci: 92, Eng: 85
# Total: 265 | Average: 88.3 | Grade: A
# --------------------------

# Use:
# __init__
# display() method
# List of objects
# Input loop for 3 students

class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
    
    def get_total(self):
        return (sum(self.marks))
    
    def get_avg(self):
        return (round(self.get_total()/3, 2))
    
    def get_grade(self):
        grade = ''
        avg = self.get_avg()
        
        if 90 <= avg <= 100:
            grade = 'A'
        elif 70 <= avg < 90:
            grade = 'B'
        elif 50 <= avg < 70:
            grade = 'C'
        else:
            grade = 'D'
        
        return grade
    
    # display method to show the student report
    def display(self):
        print("\n------------ Student Report ------------")
        print(f'Name: {self.name}')
        print(f"Math: {self.marks[0]:6}, Sci: {self.marks[1]:6}, Eng: {self.marks[2]:6}")
        print(f"Total: {self.get_total()} | Average: {self.get_avg()} | Grade: {self.get_grade()}")
        print("----------------------------------------\n")
        

records = []
flag = 'y'
stud = 1

# input student records
while (flag == 'y' or flag == 'Y'):
    print(f'Enter the details for student {stud}.')
    stud += 1
    name = input("Student name: ")
    
    # input student marks
    marks = []
    for i in range(3):
        mark = float(input(f"Enter marks earned in subject {i + 1}: "))
        marks.append(mark)
    
    # creating object
    student = Student(name, marks)
    records.append(student)    # Add object to the list
    
    flag = input("Do you want to add more student records? (Y/N) ")
    if flag not in 'yYnN':
        raise ValueError("Error! Choose (Y/N).")


# Print student report for the student of your choice
flag2 = 'y'

while (flag2 == 'y' or flag2 == 'Y'):
    search = str(input("Which student's report do you want to print? "))
    
    for each in records:
        if each.name.lower() == search.lower():
            each.display()
    
    flag2 = input("Want to continue printing reports? (Y/N) ")
    if flag2 not in 'yYnN':
        raise ValueError("Error! Choose (Y/N).")
```

```python
# 📌 Project 2: Data Loader + Simple Analysis
# ✅ Load users.csv (from Day 9)
# ✅ Count how many users are above age 20
# ✅ Print result like:
    # Total Users: 3
    # Users above age 20: 2

# 🎯 Use:
    # Class for loading
    # Method for analysis
    # __len__

import csv

class DataLoader:
    def __init__(self, filepath):
        self.filepath = filepath
        self.data = []
    
    def load_csv(self):
        with open(self.filepath, 'r') as file:
            reader = csv.DictReader(file)
            self.data = list(reader)
        return self.data
    
    def analyze(self):
        count = 0
        users_data = self.load_csv()
        
        for each in users_data:
            if int(each["Age"]) > 26:
                count += 1
        return count
    
    def __len__(self):
        return len(self.data)   # returns number of users for the users file

loader = DataLoader('users.csv')    # object created
loader.load_csv()

print(f"Total users: {loader.__len__()}")
print(f"Users above age 26: {loader.analyze()}")
```
