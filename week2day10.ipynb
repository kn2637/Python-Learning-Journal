```python
# Day 10 - Error Handling and Debugging

# try/except basics

try:
    print(10/0)
except:
    print("Error! Cannot divide by 0!")
```

```python
# Common exceptions
# ValueError - Wrong value type
try:
    val = int(input("Enter an integer: "))
    print(f'You entered {val}.')
except ValueError:
    print("Invalid! You didn't enter an integer.")
```

```python
# Common exceptions
# ZeroDivisionError - Dividing by zero

try:
    print(int(10/0))
except ZeroDivisionError:
    print("Error! Cannot divide by 0!")
```

```python
# Common exceptions
# TypeError - Wrong data type

try:
    result = 5
    char = 'c'
    print(result + char)
except TypeError:
    print("Error! Cannot add integer and string!")
```

```python
# Common exceptions
# FileNotFoundError - File missing

try:
    with open("nonExistingFile.txt", 'r') as file:
        content = file.read()
        print(content)
except FileNotFoundError:
    print("Error! The file is missing!")
```

```python
# Common exceptions
# KeyError - Dictionary key missing

dict1 = {
    'name': 'Komal',
    'age': 20
}

try:
    print(dict1['name'])
    print(dict1['course'])
except KeyError:
    print("Error! The dictionary Key is missing!")
```

```python
# Try + Except + Else + Finally
# Finally runs no matter what and runs before the returned value

try:
    num = int(input("Enter a number: "))
except ValueError:
    print("Error! Please enter a valid number!")
else:
    print("Square: ", num ** 2)
finally:
    print("Program finished")
```

```python
# Example showing finally runs even on return

def test():
    try:
        return "Returning from test"
    finally:
        print("Program finished.")

print(test())
```

```python
# Raise Your Own Custom Errors

try:
    age = int(input("Enter your age: "))
except ValueError:
    print("Error! Incorrect age entered!")
if age < 0:
    raise ValueError("Error! Age cannot be negative!")
```

```python
# Mini Project
# 📌 Student Data Validator + Saver
    # Ask student name and marks
    # Marks must be 0–100
    # Handle wrong inputs
    # Save valid records in a JSON file

import json

student_records = []

flag = 'y'

while(flag == 'y' or flag == 'Y'):
    
    try:
        name = input("Enter student's name: ")
        marks = int(input("Enter their marks (must be between 0 and 100): "))
        
        if marks not in range(0, 101):
            raise ValueError("Invalid input! Marks must be between 0 and 100.")
        
    except ValueError as err:
        print("Error: ", err)
    
    else:
        dict1 = {
        'Name': name,
        'Marks': marks
        }
        
        student_records.append(dict1)
        
        print("Student added successfully!")
    
    flag = input("Do you want to add more records? (Y/N)")
    if flag not in 'yYnN':
        raise ValueError("Invalid Input! Choose Y or N.")

# Save the valid records to a json file
with open("validated_student_marks.json", 'w') as file:
    json.dump(student_records, file, indent = 4)
    print("✅ Valid records saved successfully!")

# Reading the saved data
with open("validated_student_marks.json", 'r') as file:
    result = json.load(file)
    for row in result:
        print(f"Student Name: {row["Name"]}\nMarks: {row['Marks']}\n")
```
